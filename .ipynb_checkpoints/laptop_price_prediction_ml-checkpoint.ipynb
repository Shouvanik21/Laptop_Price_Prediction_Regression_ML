Data Preprocessing and Data Cleaning

```python
df=df.drop(columns=["laptop_ID","Product"])
```

```python
df_one_hot_encode=pd.get_dummies(df,columns=[ 'Company','TypeName',
       'ScreenResolution', 'Cpu', 'Memory', 'Gpu', 'OpSys'],drop_first=False)
```

```python
df_one_hot_encode
```

```python
exclude_cols = ['Inches', 'Ram', 'Weight', 'Price_euros']
cols_to_convert = df_one_hot_encode.columns.drop(exclude_cols)
df_one_hot_encode[cols_to_convert] = df_one_hot_encode[cols_to_convert].astype(int)
```

```python
df_one_hot_encode
```

```python
from sklearn.preprocessing import StandardScaler
numerical_cols=['Inches', 'Ram', 'Weight', 'Price_euros']
scaler=StandardScaler()
df_one_hot_encode[numerical_cols]=scaler.fit_transform(df_one_hot_encode[numerical_cols])
```

Feature Scaling and Feature Engineering and Feature Selection

```python
from scipy.stats import chi2_contingency
import pandas as pd

# Define significance level
alpha = 0.05

# 1. Bin the target variable
df_one_hot_encode['price_bin'] = pd.qcut(df_one_hot_encode['Price_euros'], q=4, labels=False)

# Identify all your one-hot encoded columns (assuming your features are all columns except targets)
# If you have specific numeric columns you want to skip testing (like Inches, Weight), put them here
numeric_features = ['Inches', 'Weight','Ram'] 
columns_to_test = [col for col in df_one_hot_encode.columns if col not in ['Price_euros', 'price_bin'] + numeric_features]

# Run Chi-Square test for each feature column individually
chi2_results = {}
cols_to_keep = []

for col in columns_to_test:
    # Generate contingency table
    contingency = pd.crosstab(df_one_hot_encode[col], df_one_hot_encode['price_bin'])
    
    # Run the Chi-Square test
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    
    # Check significance
    if p_val < alpha:
        decision = 'Reject Null (Keep Feature)'
        cols_to_keep.append(col)
    else:
        decision = 'Fail to Reject Null (Drop Feature)'
        
    chi2_results[col] = {
        'chi2_statistic': chi2_stat,
        'p_value': p_val,
        'Decision': decision
    }

# Convert results summary to a DataFrame for review
chi2_df = pd.DataFrame(chi2_results).T.sort_values(by='p_value')

# 2. Create the final_df
# We include the selected categorical columns, your original numeric features, and the target column
final_cols = cols_to_keep + numeric_features + ['Price_euros']
final_df = df_one_hot_encode[final_cols].copy()

# Drop the temporary bin column from the original dataframe to keep it clean
df_one_hot_encode = df_one_hot_encode.drop(columns=['price_bin'])

print(f"Original Columns: {df_one_hot_encode.shape[1]}")
print(f"Dropped {len(columns_to_test) - len(cols_to_keep)} insignificant categorical columns.")
print(f"Final Data Shape: {final_df.shape}")
```
